```python
#activity 4- ploting of graph of 2 cities blr and hydrabad in the same plot of the real time data
import pandas as pd
import matplotlib.pyplot as plt

# Dataset combining Bangalore (2025 data) and Hyderabad monthly average max temperatures
weather_data = {
    "Month": [
        "Jan", "Feb", "Mar", "Apr", "May", "Jun",
        "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
    ],
    "Bangalore_Max": [
        31.6, 32.7, 32.2, 32.1, 31.8, 29.1,
        29.0, 27.8, 28.1, 29.8, 30.8, 31.0
    ],
    "Hyderabad_Max": [
        28.6, 31.8, 35.2, 37.6, 38.8, 34.4,
        30.5, 29.6, 30.1, 30.4, 28.8, 27.8
    ]
}

# Load data into a Pandas DataFrame
df = pd.DataFrame(weather_data)

# Initialize the plot size
plt.figure(figsize=(10, 6))

# Plot Bangalore using a teal line
plt.plot(
    df["Month"], df["Bangalore_Max"],
    marker="o", color="#008080", linewidth=2.5, label="Bangalore (BLR)"
)

# Plot Hyderabad using a coral line to clearly differentiate them
plt.plot(
    df["Month"], df["Hyderabad_Max"],
    marker="s", color="#FF7F50", linewidth=2.5, label="Hyderabad"
)

# Customizing chart details
plt.title("2025 Monthly Average Max Temperature Comparison", fontsize=14, fontweight="bold", pad=15)
plt.xlabel("Month", fontsize=12, labelpad=10)
plt.ylabel("Temperature (°C)", fontsize=12, labelpad=10)
plt.grid(True, linestyle="--")
plt.legend(fontsize=11, loc="upper right")

# Display the line chart
plt.tight_layout()
plt.show()
```

```python
# Simple Weather Data Visualizer
import pandas as pd
import matplotlib.pyplot as plt

# Import the dataset from a CSV file
# Make sure 'weather_data.csv' is saved in your script's directory
df = pd.read_csv('weather_data.csv')

# Create the plot window
plt.figure(figsize=(10, 6))

# Plot Bangalore using a teal line
plt.plot(
    df['Month'], df['Bangalore_Max'],
    marker='o', color='#008080', linewidth=2, label='Bangalore (BLR)'
)

# Plot Hyderabad using a coral line
plt.plot(
    df['Month'], df['Hyderabad_Max'],
    marker='s', color='#FF7F50', linewidth=2, label='Hyderabad (HYD)'
)

# Chart layout customizations
plt.title('Monthly Average Max Temperature Comparison (2025)', fontsize=14, fontweight='bold')
plt.xlabel('Month', fontsize=12)
plt.ylabel('Temperature (°C)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()

# Display the chart
plt.tight_layout()
plt.show()
```
